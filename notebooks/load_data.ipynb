```python
from app.population_generator import PopulationGeneratorWithHotStart
from app.collect_prices import collect_prices
import pandas as pd
import itertools
```

```python
df = collect_prices().dropna()
```

```python
gen = PopulationGeneratorWithHotStart()
```

```python
all_individuals = gen.initial(df)

all_assets = [i.assets for i in all_individuals]
df_population = pd.DataFrame(all_assets)
```

```python
df_population.tail()
df = df_population.copy()
```

```python
# bin_labels = ['Faixa 1', 'Faixa 2', 'Faixa 3', 'Faixa 4', 'Faixa 5', 'Faixa 6', 'Faixa 7', 'Faixa 8', 'Faixa 9']
bin_labels = ['Faixa 1', 'Faixa 2', 'Faixa 3', 'Faixa 4', 'Faixa 5']
for col in df.columns:
    df[f'{col}_bin'] = pd.cut(df[col],
                               bins=len(bin_labels),
                               labels=bin_labels)

df.filter(like='_bin').head()
```

```python
df[df['TLT_bin'] == 'Faixa 1']['TLT']
```

```python
df.drop(columns=df_population.columns, inplace=True)
```

```python
df['IAU_bin'].value_counts()
```

```python
for col in df.columns:
    print(df[col].value_counts())
```

```python
all_combinations_itertools = itertools.product(bin_labels, repeat=len(df.columns))
df_combinations = pd.DataFrame(all_combinations_itertools, columns=df.columns)
```

```python
print(f"\n--- Total de Combinações Possíveis: {len(df_combinations)} ---")
print("As 5 primeiras combinações possíveis:")
df_combinations.head()
```

```python
df_exists_combinations = df[df.columns].drop_duplicates()

print(f"\n--- Total de Combinações Existentes no DataFrame: {len(df_exists_combinations)} ---")
df_exists_combinations.head()
```

```python
on_join = ['TLT_bin', 'SMAL11.SA_bin', 'BIL_bin', 'XFIX11.SA_bin',
               'BOVA11.SA_bin', 'IMAB11.SA_bin', 'IAU_bin', 'IVV_bin', 'BTC-USD_bin']

df_merged = df_combinations.merge(df_exists_combinations, on=on_join, how='left', indicator=True)

df_pending_combinations = df_merged[df_merged['_merge'] == 'left_only']

df_pending_combinations = df_pending_combinations.drop(columns='_merge')


print(f"\n--- Total de Combinações Faltantes: {len(df_pending_combinations)} ---")
print("As 10 primeiras combinações faltantes:")
df_pending_combinations.head(10)
```
